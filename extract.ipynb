```python
page= reader.pages[199]
text = page.extract_text(extraction_mode="layout")
print(text)
```

```python
pattern = re.compile(r"""
    ^\s*                          # start of line, optional spaces
    (?P<municipality>[A-Za-z\-\s]+?)  # Municipality name (lazy to stop at ZIP)
    \s+                            # spaces
    \d{4}                          # ZIP code (4 digits)
    \s+                            # spaces
    (?P<mayor>[A-Z\.\s\-]+?)       # Mayor name (lazy)
    \s{2,}                         # 2 or more spaces as separator
    (?P<vice_mayor>[A-Z\.\s\-]+)  # Vice Mayor name
    $
""", re.VERBOSE | re.MULTILINE)
```

```python
data = []
for match in pattern.finditer(text):
    re_muni_data = match.groupdict()
    data.append({
        "name": re_muni_data['mayor'].lower(),
        "location": re_muni_data['municipality'],
        "job_title": "mayor",
        "district": None
    })
    data.append({
        "name": re_muni_data['vice_mayor'].lower(),
        "location": re_muni_data['municipality'],
        "job_title": "vice-mayor",
        "district": None
    })
```

```python
muni_pages = [i for i in range(199, 254)]
muni_data = []
for muni_page in muni_pages:
    page = reader.pages[muni_page]
    text = page.extract_text(extraction_mode="layout")
    data = []
    for match in pattern.finditer(text):
        re_muni_data = match.groupdict()
        data.append({
            "name": re_muni_data['mayor'].lower(),
            "location": re_muni_data['municipality'],
            "job_title": "mayor",
            "district": None
        })
        data.append({
            "name": re_muni_data['vice_mayor'].lower(),
            "location": re_muni_data['municipality'],
            "job_title": "vice-mayor",
            "district": None
        })
        #print(data)
    muni_data.extend(data)
```
